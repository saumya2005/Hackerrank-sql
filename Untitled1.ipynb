```python

COMMON_MPINS = [
    '1234', '1111', '0000', '1212', '7777', '9999',
    '1004', '2000', '4444', '2222', '6969', '1122'
]

def is_commonly_used_mpin(mpin):
    if len(mpin) != 4 or not mpin.isdigit():
        return "Invalid MPIN. Please enter a 4-digit number."
    if mpin in COMMON_MPINS:
        return f"{mpin} is a commonly used MPIN (WEAK)"
    else:
        return f"{mpin} is not a commonly used MPIN (could be STRONG)"
```

```python

def get_demographic_variants(date_str):
    """
    Takes a date in DD-MM-YYYY format and returns possible 4-digit combinations
    """
    dd, mm, yyyy = date_str.split('-')
    yy = yyyy[-2:]

    return [
        dd + mm, mm + dd,  # DDMM, MMDD
        yy + mm, mm + yy,  # YYMM, MMYY
        dd + yy, yy + dd   # DDYY, YYDD
    ]

def check_mpin_strength(mpin, dob_self, dob_spouse, anniversary):
    if len(mpin) != 4 or not mpin.isdigit():
        return "Invalid MPIN. Please enter a 4-digit number."

    demographic_patterns = (
        get_demographic_variants(dob_self) +
        get_demographic_variants(dob_spouse) +
        get_demographic_variants(anniversary)
    )

    if mpin in COMMON_MPINS or mpin in demographic_patterns:
        return f"{mpin} is WEAK due to demographic or common pattern."
    else:
        return f"{mpin} is STRONG."

```

```python


def check_mpin_with_reasons(mpin, dob_self, dob_spouse, anniversary):
    reasons = []

    if len(mpin) != 4 or not mpin.isdigit():
        return {
            "strength": "INVALID",
            "reasons": ["INVALID_INPUT"]
        }

    if mpin in COMMON_MPINS:
        reasons.append("COMMONLY_USED")

    if mpin in get_demographic_variants(dob_self):
        reasons.append("DEMOGRAPHIC_DOB_SELF")
    if mpin in get_demographic_variants(dob_spouse):
        reasons.append("DEMOGRAPHIC_DOB_SPOUSE")
    if mpin in get_demographic_variants(anniversary):
        reasons.append("DEMOGRAPHIC_ANNIVERSARY")

    if reasons:
        return {
            "strength": "WEAK",
            "reasons": reasons
        }
    else:
        return {
            "strength": "STRONG",
            "reasons": []
        }

```

```python

COMMON_MPINS_6 = [
    '123456', '111111', '000000', '121212', '654321', '112233'
]

def get_demographic_variants_6(date_str):
    """
    Given a date in DD-MM-YYYY format, returns common 6-digit variants.
    We'll use these two variants:
    - dd + mm + last two digits of year (ddmmyy)
    - mm + dd + last two digits of year (mmddyy)
    """
    dd, mm, yyyy = date_str.split('-')
    yy = yyyy[-2:]
    return [dd + mm + yy, mm + dd + yy]

def check_mpin_6_with_reasons(mpin, dob_self, dob_spouse, anniversary):
    reasons = []

    if len(mpin) != 6 or not mpin.isdigit():
        return {
            "strength": "INVALID",
            "reasons": ["INVALID_INPUT"]
        }

    if mpin in COMMON_MPINS_6:
        reasons.append("COMMONLY_USED")


    if mpin in get_demographic_variants_6(dob_self):
        reasons.append("DEMOGRAPHIC_DOB_SELF")
    if mpin in get_demographic_variants_6(dob_spouse):
        reasons.append("DEMOGRAPHIC_DOB_SPOUSE")
    if mpin in get_demographic_variants_6(anniversary):
        reasons.append("DEMOGRAPHIC_ANNIVERSARY")

    return {
        "strength": "WEAK" if reasons else "STRONG",
        "reasons": reasons
    }


mpin_6 = '020198'
dob_self = '02-01-1998'
dob_spouse = '14-02-1997'
anniversary = '25-12-2019'

print(check_mpin_6_with_reasons(mpin_6, dob_self, dob_spouse, anniversary))
```
